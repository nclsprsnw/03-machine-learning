## <a id='toc1_5_'></a>[Outliers analysis](#toc0_)

```python
def remove_outliers_3sigma(df, columns_to_check):
    """
    Removes outliers from specified numerical columns in a DataFrame
    using the mean +/- 3 standard deviations rule.

    Args:
        df (pd.DataFrame): The input DataFrame.
        columns_to_check (list): A list of column names (strings) to check for outliers.

    Returns:
        pd.DataFrame: A new DataFrame with outliers removed from the specified columns.
                      Returns a copy to avoid SettingWithCopyWarning.
    """
    # Create a copy of the DataFrame to avoid modifying the original and
    # to prevent SettingWithCopyWarning issues when filtering.
    df_cleaned = df.copy()
    initial_rows = len(df_cleaned)
    rows_removed_total = 0

    print(f"Starting outlier removal for {len(columns_to_check)} columns...")

    for col in columns_to_check:
        if col in df_cleaned.columns and pd.api.types.is_numeric_dtype(df_cleaned[col]):
            mean_val = df_cleaned[col].mean()
            std_dev = df_cleaned[col].std()

            lower_bound = mean_val - (3 * std_dev)
            upper_bound = mean_val + (3 * std_dev)

            # Filter out rows where the column value is outside the [lower_bound, upper_bound] range
            # Use boolean indexing to keep only non-outliers
            rows_before_filter = len(df_cleaned)
            df_cleaned = df_cleaned[
                (df_cleaned[col] >= lower_bound) & (df_cleaned[col] <= upper_bound)
            ]
            rows_after_filter = len(df_cleaned)
            removed_in_col = rows_before_filter - rows_after_filter
            rows_removed_total += removed_in_col
            print(f"  - Column '{col}': Removed {removed_in_col} outliers (values outside [{lower_bound:.2f}, {upper_bound:.2f}])")
        else:
            print(f"  - Warning: Column '{col}' not found or is not numeric. Skipping outlier check for this column.")

    final_rows = len(df_cleaned)
    print(f"\nOutlier removal complete.")
    print(f"Initial rows: {initial_rows}")
    print(f"Rows after removing outliers: {final_rows}")
    print(f"Total rows removed: {initial_rows - final_rows}")

    return df_cleaned

cols_to_clean = ['Temperature', 'Fuel_Price', 'CPI', 'Unemployment']
df_cleaned_no_outliers = remove_outliers_3sigma(df_cleaned_target, cols_to_clean)
df_cleaned_no_outliers.shape
```

### <a id='toc1_6_1_'></a>[Weekly sales](#toc0_)

## <a id='toc1_7_'></a>[Feature engineering (with the dates)](#toc0_)

### <a id='toc1_7_1_'></a>[Decompose the date](#toc0_)

```python
df_cleaned_target['Date'] = pd.to_datetime(df_cleaned_target['Date'], format='%d-%m-%Y')
df_cleaned_target['Day_Of_Week'] = df_cleaned_target['Date'].dt.day_of_week
df_cleaned_target['Quarter'] = df_cleaned_target['Date'].dt.quarter
df_cleaned_target['Month'] = df_cleaned_target['Date'].dt.month
df_cleaned_target['Year'] = df_cleaned_target['Date'].dt.year
df_cleaned_target['Day_Of_Year'] = df_cleaned_target['Date'].dt.day_of_year
df_cleaned_target['Day_Of_Month'] = df_cleaned_target['Date'].dt.day
df_cleaned_target['Week_Of_Year'] = df_cleaned_target['Date'].dt.isocalendar().week
```

### <a id='toc1_7_2_'></a>[Display the trends across times](#toc0_)

```python
plot_configs = {
    'Day_Of_Week': {'type': 'box', 'title': 'Weekly Sales by Day of Week'},
    'Quarter': {'type': 'box', 'title': 'Weekly Sales by Quarter'},
    'Month': {'type': 'box', 'title': 'Weekly Sales by Month'},
    'Year': {'type': 'box', 'title': 'Weekly Sales by Year'},
    'Week_Of_Year': {'type': 'box', 'title': 'Weekly Sales by Week of Year'},
    'Day_Of_Year': {'type': 'scatter', 'title': 'Weekly Sales vs. Day of Year (Trendline)'},
    'Day_Of_Month': {'type': 'scatter', 'title': 'Weekly Sales vs. Day of Month (Trendline)'},
}

num_plots = len(plot_configs)
nrows = (num_plots + 1) // 2 
ncols = 2

# Create a figure with subplots. Subplot titles are used for individual plot headings.
fig = make_subplots(rows=nrows, cols=ncols,
                    subplot_titles=[config['title'] for config in plot_configs.values()])
# Populate the subplots with Plotly Express figures
for i, (feature, config) in enumerate(plot_configs.items()):
    row = (i // ncols) + 1 # Determine current row for subplot
    col = (i % ncols) + 1  # Determine current column for subplot

    if config['type'] == 'box':
        # Create a box plot using Plotly Express
        px_fig = px.box(df_cleaned_target, x=feature, y='Weekly_Sales',
                        color_discrete_sequence=px.colors.qualitative.Plotly) # Example color palette
    elif config['type'] == 'scatter':
        # Create a scatter plot with an OLS (Ordinary Least Squares) trendline
        px_fig = px.scatter(df_cleaned_target, x=feature, y='Weekly_Sales', trendline='ols',
                            opacity=0.6,
                            color_discrete_sequence=px.colors.qualitative.Plotly)

    # Add each trace (data series) from the Plotly Express figure to the main subplot figure
    for trace in px_fig.data:
        fig.add_trace(trace, row=row, col=col)

    # Update axes titles for clarity
    fig.update_xaxes(title_text=feature, row=row, col=col)
    fig.update_yaxes(title_text='Weekly Sales', row=row, col=col)

# --- 5. Final layout adjustments for the combined figure ---
fig.update_layout(
    title_text="<b>Weekly Sales Trends Across Different Time Features</b>",
    title_x=0.5, # Center the main title
    height=nrows * 450, # Adjust overall figure height for better spacing
    width=1200,
    showlegend=False,    # Legend often not necessary for these individual plots
    template="plotly_white" # A clean theme
)

fig.show()
```
